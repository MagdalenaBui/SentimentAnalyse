## 1. Read PDF and extract texts

```python
import pdfplumber

def extract_pages(pdf_path):
    pages_text = []
    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            # Mehrspaltiges Layout: Seite in linke/rechte Hälfte teilen
            width = page.width
            left = page.crop((0, 0, width/2, page.height))
            right = page.crop((width/2, 0, width, page.height))
            text = (left.extract_text() or "") + "\n" + (right.extract_text() or "")
            pages_text.append(text)
    return pages_text

pages = extract_pages("shanghaijewishch00unseExample.pdf")
full_text = "\n\n".join(pages)
```
